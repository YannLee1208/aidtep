```python
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1 import make_axes_locatable
import matplotlib.colors as mcolors

obs = interpolation[100]
xs = [0, 10, 20, 30, 40, 50, 60, 70, 80]
ys = [0, 10, 20, 30, 40, 50, 60, 70, 80]
xy = [(i, j) for i in xs for j in ys]

# Create a figure and axis
fig, ax = plt.subplots()

# Use correct normalization for the colormap
norm = mcolors.Normalize(vmin=obs.min(), vmax=obs.max())

# Display the image with the correct normalization
im = ax.imshow(obs, norm=norm)

# Add the scatter plot on the same axis without affecting the image scaling
ax.scatter([i for (i, j) in xy], [j for (i, j) in xy], label="sensor")

# Add a legend
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

# Create a divider for the current axis
divider = make_axes_locatable(ax)

# Append a new axis for the colorbar
cax = divider.append_axes("bottom", size="5%", pad=0.1)

# Add the colorbar with the correct normalization
plt.colorbar(im, cax=cax, orientation="horizontal")

# Save the figure if needed
plt.savefig("interpolation_with_sensor.pdf",  bbox_inches='tight')
# plt.savefig("interpolation.png")

plt.show()
```
